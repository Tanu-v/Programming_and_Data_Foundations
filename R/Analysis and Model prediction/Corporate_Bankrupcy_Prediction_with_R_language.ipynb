#Modeling

## Logistic regression model     

 glm()

```r
lr = glm(Bankruptcy ~ ., data = train, family = "binomial" )
mo1 = predict(lr, newdata = test, type = "response")
n = as.factor(ifelse(mo1 > 0.5, "1", "0"))
table(predicted = n, actual = test$Bankruptcy)
```

```r
dim(test)
```

##Decision Tree model

 rpart()       #modeling

 rpart.plot()     #for plotting

```r
ds = rpart(Bankruptcy ~ .,data = train,method = "class")
rpart.plot(ds)
mo2 = predict(ds, newdata =test,type = "class")
p = predict(ds, newdata = test,type = "prob")[,1]        #type = "prob", probability , it can add more possibility
table(predicted = mo2, actual = test$Bankruptcy)
```

##Random Forest

** randomForest()

** varImpPlot()   #is a built-in visualization function from the randomForest package in R

#more additional functions are used with it


```r
set.seed(123)    #used for reducing ovefitting
rf = randomForest(Bankruptcy ~ .,data = train,ntree = 500,importance = TRUE)    #here ntree = no of decision tree
mo3 = predict(rf, newdata = test, type = "class")    #if we use type = "class" we need to use two predict() in different variabls
p1 = predict(rf, newdata = test, type ="prob")[,1]
table(predicted = mo3,actual = test$Bankruptcy)
```

```r
importance(rf)      #You must set importance = TRUE inside the randomForest() function when training the model so R computes all importance metrics.

varImpPlot(rf)
```

##SVM Support Vector Machine
svm()



```r
sm = svm(Bankruptcy ~ .,data = train, probability = TRUE)
mo4 = predict(sm, newdata = test,probability =TRUE)
p2 = attr(mo4,"probabilities")[,1]
table(predicted =mo4,actual = test$Bankruptcy)
```

##ROC plot AND AUC calculation

```r
roc_logic = roc(test$Bankruptcy,mo1,quiet = TRUE)
roc_dec   = roc(test$Bankruptcy,p,quiet = TRUE)
roc_rf    = roc(test$Bankruptcy,p1,quiet = TRUE)
roc_svm   = roc(test$Bankruptcy,p2,quiet = TRUE)

plot(roc_logic,col="red",main = "ROC CURVES - ALL MODES")
plot(roc_dec,col="green",add = TRUE)
plot(roc_rf,col="pink",add = TRUE)
plot(roc_svm,col="blue",add = TRUE)
legend("bottomright",legend = c("Logical Regression","Decision Tree","Random Forest","Support Vector Machine(SVM)"),col =c("red","green","pink","blue"),lwd=2)
```

##MODEL COMPARISON TABLE

```r
com = function(model_name,actual,predicted_class,predicted_prob){
  cm = caret::confusionMatrix(predicted_class,actual,positive="1")
  cm1 = pROC::roc(actual,predicted_prob,quiet = TRUE)
  data.frame(
    Model = model_name,
    Accuracy = round(cm$overall['Accuracy'],2),              #2 is taken because there are two probability in Bankrupcy 0,1
    Precision = round(cm$byClass['Pos Pred Value'],2),       #2 is taken because there are two probability in Bankrupcy 0,1
    Recall = round(cm$byClass['Sensitivity'],2),             #2 is taken because there are two probability in Bankrupcy 0,1
    F1_score = round(cm$byClass['F1'],2),                    #2 is taken because there are two probability in Bankrupcy 0,1
    AUC = round(as.numeric(pROC::auc(cm1)),2),               #2 is taken because there are two probability in Bankrupcy 0,1
    row.names = NULL,
    check.names = FALSE
  )
}

result = rbind(
  com("Logistic Regression",test$Bankruptcy,n,mo1),
  com("Decision Tree",test$Bankruptcy,mo2,p),
  com("Random Forest",test$Bankruptcy,mo3,p1),
  com("SVM",test$Bankruptcy,mo4,p2)
)

print(result)
kable(result,format = "pipe",align = c("l",rep("r",5)))
```

###Random Forest is the best overall model according to my testing based on these numbers. It achieves the highest Accuracy (0.82), F1_score (0.63), and AUC (0.87) among all four options.

###Why Random Forest Wins Overall

• Best Separation (AUC = 0.87): It does the best job of telling  two groups apart across all cutoff levels.

• Best Overall Balance (F1_score = 0.63): It keeps the strongest balance between catching true cases and avoiding false alarms.

• Highest Accuracy (0.82): It makes the most total correct predictions, tying with SVM while beating it on almost every other metric.
